```python
import requests
import pandas as pd

url = "https://openlibrary.org/search.json"

params = {
    "q": "data analytics",
    "limit": 500,
    "fields": "key,title,author_name,first_publish_year,"
              "subject,isbn,language,publisher,edition_count,"
              "cover_i,publish_date"
}

headers = {
    "User-Agent": "BookAnalyticsProject (aakritigharti365@gmail.com)"
}

response = requests.get(
    url,
    params=params,
    headers=headers,
    timeout=30
)

response.raise_for_status()

data = response.json()

books = data["docs"]

print("Total books fetched:", len(books))
print("Total books found:", data.get("numFound"))
```

```python
recent_books = clean_df[clean_df["first_publication_year"] > 2015]
print(recent_books[["title", "primary_author", "first_publication_year"]].head(10))
```

```python
high_edition_books = clean_df[clean_df["total_editions"] > 10]
print(high_edition_books[["title", "primary_author", "total_editions"]])
```

```python
yearly_analysis = (
    clean_df.dropna(subset=["first_publication_year"])
      .groupby("first_publication_year")
      .agg(
          total_books=("book_key", "count"),
          average_editions=("total_editions", "mean")
      )
      .reset_index()
      .sort_values("first_publication_year")
)

yearly_analysis["average_editions"] = yearly_analysis["average_editions"].round(2)
print(yearly_analysis)
```

```python
subject_analysis = (
    clean_df[clean_df["primary_subject"] != "Unknown"]
      .groupby("primary_subject")
      .agg(total_books=("book_key", "count"))
      .reset_index()
      .sort_values("total_books", ascending=False)
      .head(10)
)
print(subject_analysis)
```

```python
edition_statistics = clean_df["total_editions"].agg( ["count", "mean", "median", "min", "max", "sum"])
print(edition_statistics)
```

```python
def publication_category(year):
    if pd.isna(year):
        return "Unknown"

    elif year < 1950:
        return "Before 1950"

    elif year < 2000:
        return "1950-1999"

    elif year < 2010:
        return "2000-2009"

    elif year < 2020:
        return "2010-2019"

    else:
        return "2020 onwards"
clean_df["publication_period"] =clean_df["first_publication_year"].apply(publication_category)
```

```python
print(clean_df[["title","publication_period"]])
```

```python
clean_df["edition_category"] =clean_df["total_editions"].apply(
    lambda x: (
        "Single Edition" if pd.notna(x) and x == 1
        else "2-5 Editions" if pd.notna(x) and 2 <= x <= 5
        else "6-10 Editions" if pd.notna(x) and 6 <= x <= 10
        else "More than 10 Editions" if pd.notna(x) and x > 10
        else "Unknown"
    )
)
```

```python
print(clean_df[["title","edition_category"]])
```

```python
author_counts = (
    clean_df[clean_df["primary_author"] != "Unknown"]
    .groupby("primary_author")
    .agg(total_books_by_author=("book_key", "count"))
    .reset_index()
)
books_with_author_stats = pd.merge(clean_df,author_counts,on="primary_author",how="left")
print(
    books_with_author_stats[["title", "primary_author", "total_books_by_author"]]
)
```
